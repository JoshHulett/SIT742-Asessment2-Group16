### Q1. General naive-lag forecast table generator [5 marks]

Implement `generate_naive_forecast_wide(historical_actual_wide, cutoff_label, forecast_months, required_destinations=None, lag=1, date_column="Date")`.

Requirements: accept a wide actual table; return a pandas DataFrame with exactly one row per requested month; use all destination columns when `required_destinations=None`; otherwise forecast exactly the supplied destinations; use only values available at or before `cutoff_label`; support positive integer `lag`, especially `lag=1` and `lag=12`; support arbitrary forecast months and all destinations; compute recursive lag forecasts in chronological month order; avoid string subtraction for month labels.

Return rows in the supplied `forecast_months` order. Return columns in the exact order `date_column`, followed by the inferred or supplied destination order. Do not return an index column, `model_label`, diagnostics, or other extra columns.

Lag rule: for forecast month `m`, use source month `m - lag`. If the source month is after the cutoff but has already been forecast, use that generated forecast recursively. If no source value is available, return `NaN` or raise a clear error.

`generate_naive_forecast_wide` takes a wide monthly table of actual value and returns a wide forecast DataFrame in the structure of `Date` then the destination columns. #### Optional month-label helper example

This optional helper example only shows one way to convert labels such as `2023M03` into a month-aware representation. You may implement your own helper inside Q1. Do not use string subtraction for month arithmetic.


````python
def month_label_to_period_example(month_label):
    """Optional example: convert a YYYYMmm label to a pandas monthly Period."""
    return pd.Period(str(month_label).replace("M", "-"), freq="M")


def period_to_month_label_example(period):
    """Optional example: convert a pandas monthly Period back to YYYYMmm."""
    return f"{period.year}M{period.month:02d}"

# Example only:
# month_label_to_period_example("2024M01") - 12
````

````python
def generate_naive_forecast_wide(
    historical_actual_wide,
    cutoff_label,
    forecast_months,
    required_destinations=None,
    lag=1,
    date_column="Date",
):
    """Return the required wide-format naive-lag forecast DataFrame.

    Return exactly one row per requested month in the supplied forecast_months
    order. Return columns as date_column followed by the inferred or supplied
    destination order, with no extra columns. If required_destinations is None,
    use all columns except date_column in historical_actual_wide.
    """
    if isinstance(lag, bool) or not isinstance(lag, (int, np.integer)) or lag < 1:
        raise ValueError(f"Lag must be a positive integer, got {lag!r}")
    
    if required_destinations is None:
        destinations = [col for col in historical_actual_wide.columns if col != date_column]
    else:
        destinations = list(required_destinations)
    
    # Convert cutoff period to proper Period
    cutoff_period = month_label_to_period_example(cutoff_label)

    # Add Period column and only keep rows before cutoff
    historical = historical_actual_wide.copy()
    historical["period"] = historical[date_column].apply(month_label_to_period_example)
    historical = historical[historical["period"] <= cutoff_period]

    # Build lookup table for period -> {destination: value}
    known_values = {}
    for _, row in historical.iterrows():
        known_values[row["period"]] = {dest: row[dest] for dest in destinations}

    forecast_values = {}

    target_periods = [month_label_to_period_example(month) for month in forecast_months]

    sorted_periods = sorted(set(target_periods))

    for target_period in sorted_periods:
        source_period = target_period - lag

        row_values = {}
        for dest in destinations:
            if source_period in known_values:
                # User observed value for the month
                row_values[dest] = known_values[source_period][dest]
            elif source_period in forecast_values:
                # Use forecasted value for this month
                row_values[dest] = forecast_values[source_period][dest]
            else:
                # No data, return NaN
                row_values[dest] = np.nan
        
        forecast_values[target_period] = row_values
    
    # Build output table
    output_rows = []
    for month_label, target_period in zip(forecast_months, target_periods):
        row = {date_column: month_label}
        row.update(forecast_values[target_period])
        output_rows.append(row)
    
    result = pd.DataFrame(output_rows, columns=[date_column] + destinations)
    return result
````

### Q2. Forecast and actual table validation/alignment [5 marks]

Duplicate counts count occurrences after the first occurrence.

**Required answer markers:** Do not delete, rename, edit, move, or reorder the Q2 START/END marker comments below. Write your response between them.

<!-- A2:ANSWER:Q2:START -->
`validate_forecast_actual_wide` returns a 20-key Python dictionary `Q2_AUDIT_FIELDS` in the required order. The function has two modes:

Forecast-only mode is called by `actual_wide_df=None`. This checks the missing or duplicated required months, missing destination columns or unexpected columns, duplicate Date values, compare row count with the expected row count, and the missing, non-numeric, and non-finite values for each destination column. `is_valid` is returned true only if all the above structural and value checks pass. `can_align` and the keys only relevant for actual values are set to `None`.

Forecast-vs-actual mode performs the same checks in addition to the same checks on the `actual_wide_df`. `actual_wide_df` is first made into a subset with only the `required_months`. `is_align` is returned true if both actual and forecast data structures can be structurally merged. `is_valid` is returned true if both actual and forecast data structures pass the value-quality checks, indicating that they are clean.

````python
def validate_forecast_actual_wide(
    forecast_wide_df,
    actual_wide_df=None,
    required_months=None,
    required_destinations=None,
    date_column="Date",
):
    """Return the fixed Q2 audit dictionary for wide-table validation.

    Populate every key in Q2_AUDIT_FIELDS. Forecast-only mode uses None for
    can_align and every actual-related field. Forecast-vs-actual mode subsets
    actual_wide_df to required_months before reporting actual_row_count and
    checking structural alignment.
    """
    def is_number(value):
        return isinstance(value, (int, float, np.integer, np.floating)) and not isinstance(value, bool)

    forecast_row_count = len(forecast_wide_df)
    expected_row_count = len(required_months) if required_months is not None else None

    # Destination columns
    if required_destinations is not None:
        missing_destinations_in_forecast = [
            dest for dest in required_destinations if dest not in forecast_wide_df.columns
        ]
        extra_columns_in_forecast = [
            col for col in forecast_wide_df.columns
            if col != date_column and col not in required_destinations
        ]
        forecast_destination_columns = [
            dest for dest in required_destinations if dest in forecast_wide_df.columns
        ]
    else:
        missing_destinations_in_forecast = []
        extra_columns_in_forecast = []
        forecast_destination_columns = [col for col in forecast_wide_df.columns if col != date_column]
    
    # Missing/extra months
    has_date_column = date_column in forecast_wide_df.columns
    forecast_dates_present = list(dict.fromkeys(forecast_wide_df[date_column]) if has_date_column else [])
    if required_months is not None:
        missing_months_in_forecast = [m for m in required_months if m not in forecast_dates_present]
        extra_months_in_forecast = [m for m in forecast_dates_present if m not in required_months]
    else:
        missing_months_in_forecast = []
        extra_months_in_forecast = []
    
    # Duplicate dates
    duplicate_forecast_dates = 0
    for count in forecast_wide_df[date_column].value_counts():
        if count > 1:
            duplicate_forecast_dates += count - 1
    
    # Counting missing, non-numeric, non-finite values in the destination columns
    missing_forecast_value_count = 0
    nonnumeric_forecast_value_count = 0
    nonfinite_forecast_value_count = 0
    for col in forecast_destination_columns:
        for value in forecast_wide_df[col]:
            if pd.isna(value):
                missing_forecast_value_count += 1
            elif not is_number(value):
                nonnumeric_forecast_value_count += 1
            elif not np.isfinite(value):
                nonfinite_forecast_value_count += 1

    # Structurally valid if nothin is missing, extra, or duplicated
    forecast_structure_valid = (
        missing_months_in_forecast == []
        and extra_months_in_forecast == []
        and missing_destinations_in_forecast == []
        and extra_columns_in_forecast == []
        and duplicate_forecast_dates == 0
        and (expected_row_count is None or forecast_row_count == expected_row_count)
    )
    # Values valid if no missing, nonnumeric, or nonfinite values
    forecast_values_valid = (
        missing_forecast_value_count == 0
        and nonnumeric_forecast_value_count == 0
        and nonfinite_forecast_value_count == 0
    )

    if actual_wide_df is None:
        # Forecast-only mode, nothing to compare
        can_align = None
        actual_row_count = None
        missing_months_in_actual = None
        extra_months_in_actual_source = None
        missing_destinations_in_actual = None
        duplicate_actual_dates = None
        missing_actual_value_count = None
        nonnumeric_actual_value_count = None
        nonfinite_actual_value_count = None

        is_valid = forecast_structure_valid and forecast_values_valid

    else:
        # Mirror forecast operations for the actual source
        # Destination columns
        if required_destinations is not None:
            missing_destinations_in_actual = [
                dest for dest in required_destinations if dest not in actual_wide_df.columns
            ]
            actual_destination_columns = [
                dest for dest in required_destinations if dest in actual_wide_df.columns
            ]
        else:
            missing_destinations_in_actual = []
            actual_destination_columns = [col for col in actual_wide_df.columns if col != date_column]

        # Find extra months in actual
        actual_dates_present = list(dict.fromkeys(actual_wide_df[date_column]))
        if required_months is not None:
            extra_months_in_actual_source = [m for m in actual_dates_present if m not in required_months]
            actual_subset = actual_wide_df[actual_wide_df[date_column].isin(required_months)]
        else:
            extra_months_in_actual_source = []
            actual_subset = actual_wide_df

        # Missing months in actual
        actual_row_count = len(actual_subset)
        actual_subset_dates = list(dict.fromkeys(actual_subset[date_column]))
        if required_months is not None:
            missing_months_in_actual = [m for m in required_months if m not in actual_subset_dates]
        else:
            missing_months_in_actual = []

        # Duplicate dates in actual
        duplicate_actual_dates = 0
        for count in actual_subset[date_column].value_counts():
            if count > 1:
                duplicate_actual_dates += count - 1

        # Counting missing, non-numeric, non-finite values in the destination columns
        missing_actual_value_count = 0
        nonnumeric_actual_value_count = 0
        nonfinite_actual_value_count = 0
        for col in actual_destination_columns:
            for value in actual_subset[col]:
                if pd.isna(value):
                    missing_actual_value_count += 1
                elif not is_number(value):
                    nonnumeric_actual_value_count += 1
                elif not np.isfinite(value):
                    nonfinite_actual_value_count += 1

        # Structurally valid if nothin is missing, extra, or duplicated
        actual_structure_valid = (
            missing_months_in_actual == []
            and missing_destinations_in_actual == []
            and duplicate_actual_dates == 0
        )
        # Values valid if no missing, nonnumeric, or nonfinite values
        actual_values_valid = (
            missing_actual_value_count == 0
            and nonnumeric_actual_value_count == 0
            and nonfinite_actual_value_count == 0
        )

        # Forecast and actual can be aligned if there are no missing, duplicate, or extra values between them
        can_align = (
            missing_months_in_forecast == []
            and missing_months_in_actual == []
            and missing_destinations_in_forecast == []
            and missing_destinations_in_actual == []
            and duplicate_forecast_dates == 0
            and duplicate_actual_dates == 0
        )

        # Merge is valid if the structure and values are valid and match
        is_valid = (
            forecast_structure_valid
            and forecast_values_valid
            and actual_structure_valid
            and actual_values_valid
            and can_align
        )

    # Build directory
    audit = {
        "is_valid": is_valid,
        "can_align": can_align,
        "forecast_row_count": forecast_row_count,
        "actual_row_count": actual_row_count,
        "expected_row_count": expected_row_count,
        "missing_months_in_forecast": missing_months_in_forecast,
        "missing_months_in_actual": missing_months_in_actual,
        "extra_months_in_forecast": extra_months_in_forecast,
        "extra_months_in_actual_source": extra_months_in_actual_source,
        "missing_destinations_in_forecast": missing_destinations_in_forecast,
        "missing_destinations_in_actual": missing_destinations_in_actual,
        "extra_columns_in_forecast": extra_columns_in_forecast,
        "duplicate_forecast_dates": duplicate_forecast_dates,
        "duplicate_actual_dates": duplicate_actual_dates,
        "missing_forecast_value_count": missing_forecast_value_count,
        "missing_actual_value_count": missing_actual_value_count,
        "nonnumeric_forecast_value_count": nonnumeric_forecast_value_count,
        "nonnumeric_actual_value_count": nonnumeric_actual_value_count,
        "nonfinite_forecast_value_count": nonfinite_forecast_value_count,
        "nonfinite_actual_value_count": nonfinite_actual_value_count,
    }

    return audit
````

### Q3. Forecast accuracy measures from forecast and actual tables [5 marks]

Implement `evaluate_forecast_wide(forecast_wide_df, actual_wide_df, training_actual_wide_df, start_month, end_month, required_destinations=None, naive_lag=1, date_column="Date")`.

Use the [A2 Forecast Measures Guide](SIT742-2026T2-A2-Forecast-Measures-Guide.md). Both MASE and MAPE are required.

Requirements: align forecast and actual/result tables by `Date` and destination; work destination by destination; select months from `start_month` to `end_month`; evaluate all forecast destination columns when `required_destinations=None`; compute MAE, MASE, MAPE, denominator, `n`, `n_denominator_pairs`, `mase_available`, and `denominator_warning`; keep MASE/MAPE separately identifiable; sort training actuals chronologically; use all valid lagged training pairs for each destination's MASE denominator; use `naive_lag=1` for official validation unless stated otherwise; handle missing values, nonfinite values, zero actuals for MAPE, and unavailable denominators explicitly.

Return exactly `(destination_metrics, aggregate_metrics)`. `destination_metrics` must be a pandas DataFrame with columns in `DESTINATION_METRIC_COLUMNS`. `aggregate_metrics` must be a Python dictionary with exactly the keys in `AGGREGATE_METRIC_FIELDS`.

**Required answer markers:** Do not delete, rename, edit, move, or reorder the Q3 START/END marker comments below. Write your response between them.

<!-- A2:ANSWER:Q3:START -->
`evaluate_forecast_wide` returns `destination_metrics`, `aggregate_metrics`, a DataFrame ordered by `DESTINATION_METRIC_COLUMNS` and a dictionary ordered by `AGGREGATE_METRIC_FIELDS`.

The function restricts forecast and actual tables to the `start_month` and `end_month` window through `period` comparisons, the merges the forecast and actual tables destination-by-destination on the evaluation period. Only paired rows impact the evaluation measures. MAE is the mean absolute error over those pairs. MAPE is computed for rows with non-zero actuals, becoming `NaN` if no valid rows remain. MASE and MAPE are calculated and kept separate.

The MASE denominator is specified per destination and is training-only. `training_actual_wide_df` is sorted chronologically by Date. The denominator for a destination is the mean absolute lag change over the destination's consecutive training pairs. `n_denominator_pairs` and `denominator` are recorded. `denominator` is flagged true if there are non-zero pairs, a non-finite denominator, or a zero denominator. `mase_available` is true only if there are evaluate rows, a finite MAE, and no denominator warnings. If false, `mase` is recorded as `NaN` to avoid a divide by zero.

`aggregate_metrics` records the `n_destinations`, `mean_mase`, `median_mase`, and the `mean_mape` for all the destination rows.
<!-- A2:ANSWER:Q3:END -->

````python
def evaluate_forecast_wide(
    forecast_wide_df,
    actual_wide_df,
    training_actual_wide_df,
    start_month,
    end_month,
    required_destinations=None,
    naive_lag=1,
    date_column="Date",
):
    """Return (destination_metrics, aggregate_metrics).

    destination_metrics is a DataFrame using DESTINATION_METRIC_COLUMNS.
    aggregate_metrics is a dictionary using AGGREGATE_METRIC_FIELDS.
    Sort training_actual_wide_df chronologically before computing all valid
    denominator pairs. If required_destinations is None, evaluate all forecast
    columns except date_column; otherwise evaluate exactly those destinations.
    """
    
    if isinstance(naive_lag, bool) or not isinstance(naive_lag, (int, np.integer)) or naive_lag < 1:
        raise ValueError(f"Lag must be a positive integer, got {naive_lag!r}")
    
    if required_destinations is None:
        destinations = [col for col in forecast_wide_df.columns if col != date_column]
    else:
        destinations = list(required_destinations)
        
    start_period = month_label_to_period(start_month)
    end_period = month_label_to_period(end_month)
    
    def cut_window_subset(df):
        periods = df[date_column].map(month_label_to_period)
        inside_window_period = (periods >= start_period) & (periods <= end_period)
        subset =df.loc[inside_window_period].copy()
        subset["temp_period"] = periods[inside_window_period]
        return subset
    
    aligned_forecast = cut_window_subset(forecast_wide_df)
    aligned_actual = cut_window_subset(actual_wide_df)
    
    training_sorted = training_actual_wide_df.copy()
    training_sorted["temp_period"] = training_sorted[date_column].map(month_label_to_period)
    training_sorted = training_sorted.sort_values("temp_period").drop(columns="temp_period").reset_index(drop=True)
    
    training_by_month = training_sorted.drop(columns=[date_column]).apply(pd.to_numeric, errors="coerce")
    training_by_month.index = pd.PeriodIndex(training_sorted[date_column].map(month_label_to_period), freq="M")
    
    destination_metric_rows=[]
    for destination in destinations:
        if destination in aligned_forecast.columns and destination in aligned_actual.columns:
            paired = aligned_forecast[["temp_period", destination]].merge(
                aligned_actual[["temp_period", destination]],
                on="temp_period",
                suffixes=("_forecast", "_actual")
            )
            forecast_values = pd.to_numeric(paired[f"{destination}_forecast"], errors="coerce")
            actual_values = pd.to_numeric(paired[f"{destination}_actual"], errors="coerce")
        else:
            forecast_values = pd.Series(dtype=float)
            actual_values = pd.Series(dtype=float)
            
        absolute_errors = (actual_values - forecast_values).abs().replace([np.inf, -np.inf], np.nan).dropna()
        
        n = len(absolute_errors)
        mae = absolute_errors.mean() if n else np.nan
        
        actual_value = actual_values.loc[absolute_errors.index]
        nonzero = actual_value != 0
        
        if nonzero.any():
            mape = (absolute_errors[nonzero] / actual_value[nonzero].abs()).mean() *100
        else:
            mape = np.nan
        
        if destination in training_by_month.columns:
            history = training_by_month[destination].dropna()
        else:
            history = pd.Series(dtype=float)
        earlier = history.copy()
        earlier.index = earlier.index + naive_lag
        absolute_changes = (history - earlier).abs().dropna()
        
        n_denominator_pairs = len(absolute_changes)
        denominator = absolute_changes.mean() if n_denominator_pairs else np.nan
        denominator_warning = not (n_denominator_pairs > 0 and np.isfinite(denominator) and denominator != 0)
        mase_available = bool(n > 0 and np.isfinite(mae) and not denominator_warning)
        
        destination_metric_rows.append({
            "destination": destination,
            "n": n,
            "mae": mae,
            "mase": mae / denominator if mase_available else np.nan,
            "mape": mape,
            "denominator": denominator,
            "n_denominator_pairs": n_denominator_pairs,
            "mase_available": mase_available,
            "denominator_warning": bool(denominator_warning),
        })
    
    destination_metrics = pd.DataFrame(destination_metric_rows, columns=DESTINATION_METRIC_COLUMNS)
    
    aggregate_metrics = {
        "n_destinations": len(destination_metrics),
        "mean_mase": float(destination_metrics["mase"].mean()),
        "median_mase":float(destination_metrics["mase"].median()),
        "mean_mape":float(destination_metrics["mape"].mean()), 
    }
    
    return destination_metrics, aggregate_metrics
````

### 4.2 Cutoff-safe modelling data - `modelling_data_or_feature_table`

````python
def compute_destination_seasonal_profile(history_wide, cutoff_label, destinations, level_window=12, season_years=2, date_column=FINAL_FORECAST_DATE_COLUMN):
    numeric = history_wide.drop(columns=[date_column]).apply(pd.to_numeric, errors="coerce")
    numeric.index = pd.PeriodIndex(history_wide[date_column].map(month_label_to_period), freq="M")
    history = numeric.sort_index().loc[:month_label_to_period(cutoff_label)]
    assert (history.notna().sum() >= 24).all()

    rows = []
    for destination in destinations:
        observed = history[destination].dropna().astype(float)
        calendar = observed.reindex(pd.period_range(observed.index[0], observed.index[-1], freq="M"))
        ratio = (calendar / calendar.rolling(12, center=True, min_periods=12).mean()).dropna()
        monthly = ratio.iloc[-season_years * 12:].groupby(lambda p: p.month).mean()
        seasonal = (monthly / monthly.mean()).reindex(range(1, 13)).fillna(1.0)

        rows.append({
            "destination": destination,
            "n_observations": len(observed),
            "last_level": observed.iloc[-1],
            "recent_drift_12m": calendar.diff().dropna().iloc[-level_window:].mean(),
            "seasonal_amplitude": seasonal.max() - seasonal.min(),
            **{f"seasonal_{m:02d}": seasonal[m] for m in range(1, 13)},
        })
    return pd.DataFrame(rows).set_index("destination")
````

#### 4.6.1 Seasonal Decomposition and Naive-Decomposition Blend

````python
DRIFT_DAMPING = 0.9

def build_seasonal_decomposition(history_wide, cutoff_label, forecast_months, destinations, date_column=FINAL_FORECAST_DATE_COLUMN):
    """Damped-drift extrapolation of level, reseasonalised using each destination's
    own seasonal index."""
    profile = compute_destination_seasonal_profile(history_wide, cutoff_label, destinations, date_column=date_column)
    cutoff_seasonal = profile[f"seasonal_{month_label_to_period(cutoff_label).month:02d}"]

    rows = []
    for horizon, month_label in enumerate(forecast_months, start=1):
        damping = sum(DRIFT_DAMPING ** i for i in range(1, horizon + 1))
        level = profile["last_level"] + damping * profile["recent_drift_12m"]
        month_num = month_label_to_period(month_label).month
        seasonal = profile[f"seasonal_{month_num:02d}"] / cutoff_seasonal
        forecast = (level * seasonal).clip(lower=0)
        rows.append({date_column: month_label, **forecast.reindex(destinations).to_dict()})
    return pd.DataFrame(rows, columns=[date_column] + list(destinations))


def build_naive_decomposition_blend(history_wide, cutoff_label, forecast_months, destinations, date_column=FINAL_FORECAST_DATE_COLUMN):
    """Equal-weight blend of naive_lag1 and build_seasonal_decomposition."""
    naive_wide = generate_naive_forecast_wide(history_wide, cutoff_label, forecast_months, destinations, lag=1, date_column=date_column)
    decomposition_wide = build_seasonal_decomposition(history_wide, cutoff_label, forecast_months, destinations, date_column=date_column)
    blended = naive_wide.copy()
    blended[destinations] = 0.5 * naive_wide[destinations].to_numpy(float) + 0.5 * decomposition_wide[destinations].to_numpy(float)
    return blended
````

````python
# Helper Functions
def load_cutoff_safe_history(history_wide, cutoff_label, date_column=FINAL_FORECAST_DATE_COLUMN):
    numeric = history_wide.drop(columns=[date_column]).apply(pd.to_numeric, errors="coerce")
    numeric.index = pd.PeriodIndex(history_wide[date_column].map(month_label_to_period), freq="M")
    return numeric.sort_index().loc[:month_label_to_period(cutoff_label)]

def naive_fallback_forecast(history_wide, cutoff_label, forecast_months, destination, date_column=FINAL_FORECAST_DATE_COLUMN):
    fallback = generate_naive_forecast_wide(history_wide, cutoff_label, forecast_months, required_destinations=[destination], lag=1, date_column=date_column)
    return dict(zip(forecast_months, fallback[destination]))

def wide_from_forecast_dict(forecast_by_destination, forecast_months, destinations, date_column=FINAL_FORECAST_DATE_COLUMN):
    rows = [{date_column: month, **{d: forecast_by_destination[d][month] for d in destinations}} for month in forecast_months]
    return pd.DataFrame(rows, columns=[date_column] + list(destinations))
````

#### 4.6.2 STL Damped-Trend Decomposition

````python
# STL trend extrapolation bounds. The window reaches undisrupted cycles at both
# cutoffs; the caps stop log-space trend growth compounding over the horizon.
STL_WINDOW_MONTHS = 96
STL_MAX_MONTHLY_TREND_GROWTH = 1.06
STL_MAX_HORIZON_GROWTH = 2.5

def build_stl_damped(history_wide, cutoff_label, forecast_months, destinations,
                     window=STL_WINDOW_MONTHS, damping=0.85,
                     date_column=FINAL_FORECAST_DATE_COLUMN):
    """STL damped-trend forecast on a log scale.

    STL supplies the deseasonalised trend only. The seasonal shape comes from
    compute_destination_seasonal_profile, which estimates it from undisrupted
    cycles, because STL's own seasonal component is time-varying and averaging
    it across the pandemic mixes the level break into the seasonal term. The
    trend slope is bounded per month and cumulatively over the horizon.
    """
    from statsmodels.tsa.seasonal import STL
    history = load_cutoff_safe_history(history_wide, cutoff_label, date_column)
    seasonal_profile = compute_destination_seasonal_profile(
        history_wide, cutoff_label, destinations, date_column=date_column
    )
    slope_cap = np.log(STL_MAX_MONTHLY_TREND_GROWTH)
    horizon_cap = np.log(STL_MAX_HORIZON_GROWTH)

    forecast_by_destination = {}
    for destination in destinations:
        observed = history[destination].dropna().astype(float) if destination in history.columns else pd.Series(dtype=float)
        series_window = observed.iloc[-window:] if len(observed) > window else observed
        if len(series_window) < 24:
            forecast_by_destination[destination] = naive_fallback_forecast(history_wide, cutoff_label, forecast_months, destination, date_column)
            continue

        log_series = np.log(series_window.clip(lower=1.0))
        try:
            decomposition = STL(log_series.values, period=12, robust=True).fit()
            trend = decomposition.trend
            recent_slope = float(np.clip(np.mean(np.diff(trend[-12:])), -slope_cap, slope_cap))
            last_trend_value = trend[-1]

            # trend is already deseasonalised, so no cutoff-month division here
            seasonal_by_month = {
                month: float(np.log(max(seasonal_profile.loc[destination, f"seasonal_{month:02d}"], 1e-6)))
                for month in range(1, 13)
            }

            dest_forecast = {}
            for horizon, month_label in enumerate(forecast_months, start=1):
                damping_factor = sum(damping ** i for i in range(1, horizon + 1))
                trend_growth = min(damping_factor * recent_slope, horizon_cap)
                month_num = month_label_to_period(month_label).month
                dest_forecast[month_label] = float(np.exp(last_trend_value + trend_growth + seasonal_by_month[month_num]))
            forecast_by_destination[destination] = dest_forecast
        except Exception:
            forecast_by_destination[destination] = {month: float(series_window.iloc[-1]) for month in forecast_months}

    return wide_from_forecast_dict(forecast_by_destination, forecast_months, destinations, date_column)


validation_stl_damped_96m = build_stl_damped(training_actual_wide_to_2023M02, TRAIN_END, VALIDATION_MONTHS, REQUIRED_DESTINATIONS)
MODEL_REGISTRY.append((validation_stl_damped_96m, "stl_damped_96m", "decomposition"))
````

#### 4.6.3 Bounded Recovery

````python
STABLE_YEARS_BACK_START = 8
STABLE_YEARS_BACK_END = 4
RECENT_WINDOW = 15

def build_bounded_recovery(history_wide, cutoff_label, forecast_months, destinations, date_column=FINAL_FORECAST_DATE_COLUMN):
    history = load_cutoff_safe_history(history_wide, cutoff_label, date_column)
    cutoff_period = month_label_to_period(cutoff_label)

    forecast_by_destination = {}
    for destination in destinations:
        observed = history[destination].dropna().astype(float) if destination in history.columns else pd.Series(dtype=float)
        if len(observed) < 36:
            forecast_by_destination[destination] = naive_fallback_forecast(history_wide, cutoff_label, forecast_months, destination, date_column)
            continue

        calendar = observed.reindex(pd.period_range(observed.index[0], observed.index[-1], freq="M"))
        stable_start = cutoff_period - STABLE_YEARS_BACK_START * 12
        stable_end = cutoff_period - STABLE_YEARS_BACK_END * 12
        stable_window = calendar.loc[stable_start:stable_end].dropna()
        if len(stable_window) < 24:
            stable_window = calendar.dropna().iloc[-24:]

        ratio = (stable_window / stable_window.rolling(12, center=True, min_periods=12).mean()).dropna()
        seasonal = ratio.groupby(lambda period: period.month).mean()
        seasonal = (seasonal / seasonal.mean()).reindex(range(1, 13)).fillna(1.0)
        stable_deseasonalised = stable_window / stable_window.index.map(lambda period: seasonal[period.month])
        asymptote = stable_deseasonalised.mean()

        recent = calendar.dropna().iloc[-RECENT_WINDOW:]
        recent_deseasonalised = (recent / recent.index.map(lambda period: seasonal[period.month])).values
        gaps = asymptote - recent_deseasonalised[:-1]
        diffs = np.diff(recent_deseasonalised)
        denominator = np.sum(gaps ** 2)
        speed = float(np.clip(np.sum(gaps * diffs) / denominator if denominator > 1e-9 else 0.0, 0.0, 0.9))

        level = recent_deseasonalised[-1]
        dest_forecast = {}
        for month_label in forecast_months:
            level = level + speed * (asymptote - level)
            month_num = month_label_to_period(month_label).month
            dest_forecast[month_label] = max(level * seasonal[month_num], 0.0)
        forecast_by_destination[destination] = dest_forecast

    return wide_from_forecast_dict(forecast_by_destination, forecast_months, destinations, date_column)

validation_bounded_recovery = build_bounded_recovery(training_actual_wide_to_2023M02, TRAIN_END, VALIDATION_MONTHS, REQUIRED_DESTINATIONS)
MODEL_REGISTRY.append((validation_bounded_recovery, "bounded_recovery", "bounded_growth_recovery"))
````

### 4.7 Adaptive Selection

````python
def build_adaptive_selection(model_registry, validation_metrics_so_far, forecast_months, destinations, history_wide, cutoff_label, date_column=FINAL_FORECAST_DATE_COLUMN):
    forecasts_by_label = {label: forecast_wide for forecast_wide, label, _ in model_registry}
    mase_by_destination = validation_metrics_so_far.pivot(index="destination", columns="model_label", values="mase")
    best_model_per_destination = mase_by_destination.idxmin(axis=1)
    
    rows = []
    for month_label in forecast_months:
        row = {date_column: month_label}
        for destination in destinations:
            chosen_model = best_model_per_destination[destination]
            row[destination] = forecasts_by_label[chosen_model].set_index(date_column).loc[month_label, destination]
        rows.append(row)
    combined = pd.DataFrame(rows, columns=[date_column] + list(destinations))
    guarded = apply_stability_guardrails(combined, history_wide, cutoff_label, destinations, date_column=date_column)
    return guarded, best_model_per_destination

def apply_stability_guardrails(forecast_wide, history_wide, cutoff_label, destinations, max_monthly_growth=1.8, min_monthly_growth=0.5, date_column=FINAL_FORECAST_DATE_COLUMN):
    history = load_cutoff_safe_history(history_wide, cutoff_label, date_column)
    guarded = forecast_wide.copy()
    for destination in destinations:
        observed = history[destination].dropna().astype(float) if destination in history.columns else pd.Series(dtype=float)
        previous = observed.iloc[-1] if len(observed) else np.nan
        values = guarded[destination].to_numpy(dtype=float).copy()
        for i in range(len(values)):
            if not np.isfinite(previous) or previous <= 0:
                previous = max(values[i], 0.0)
                continue
            values[i] = max(float(np.clip(values[i], previous * min_monthly_growth, previous * max_monthly_growth)), 0.0)
            previous = values[i]
        guarded[destination] = values
    return guarded

NAIVE_MODEL_LABELS = {"naive_lag1", "naive_lag12"}

def apply_plausibility_fallback(
    model_by_destination,
    forecasts_by_label,
    validation_metrics_df,
    history_wide,
    cutoff_label,
    destinations,
    date_column=FINAL_FORECAST_DATE_COLUMN,
):
    """Replace a destination's selected model when its final forecast exceeds the
    destination's all-time monthly maximum in the cutoff-safe history.

    The replacement is the highest-ranked candidate by validation MASE that is not a
    naive baseline and whose own forecast stays within that maximum. The record is
    computed from history up to cutoff_label only, so the rule is cutoff-safe."""
    history = load_cutoff_safe_history(history_wide, cutoff_label, date_column)
    guarded_by_label = {
        label: apply_stability_guardrails(
            forecast_wide, history_wide, cutoff_label, destinations, date_column=date_column
        )
        for label, forecast_wide in forecasts_by_label.items()
    }

    resolved = {}
    audit_rows = []
    for destination in destinations:
        observed = history[destination].dropna().astype(float)
        record = observed.max() if len(observed) else np.nan
        selected = model_by_destination[destination]
        selected_peak = guarded_by_label[selected][destination].max()

        ranking = (
            validation_metrics_df.loc[validation_metrics_df["destination"] == destination]
            .sort_values("mase")["model_label"]
            .tolist()
        )
        eligible = [
            label for label in ranking
            if label in guarded_by_label
            and label not in NAIVE_MODEL_LABELS
            and guarded_by_label[label][destination].max() <= record
        ]

        triggered = bool(np.isfinite(record) and selected_peak > record)
        chosen = eligible[0] if triggered and eligible else selected
        resolved[destination] = chosen

        audit_rows.append({
            "destination": destination,
            "selected_model": selected,
            "record": record,
            "selected_peak": selected_peak,
            "peak_over_record": selected_peak / record if np.isfinite(record) and record > 0 else np.nan,
            "rule_triggered": triggered,
            "fallback_model": chosen if chosen != selected else "",
            "fallback_peak": guarded_by_label[chosen][destination].max(),
        })

    return pd.Series(resolved).reindex(destinations), pd.DataFrame(audit_rows)

_adaptive_selection_metrics_parts = []
for forecast_wide, label, family in MODEL_REGISTRY:
    dest_metrics, _ = evaluate_forecast_wide(forecast_wide, validation_actual_wide_2023M03_2023M07, training_actual_wide_to_2023M02, VALIDATION_START, VALIDATION_END, required_destinations=REQUIRED_DESTINATIONS, naive_lag=1)
    _adaptive_selection_metrics_parts.append(dest_metrics.assign(model_label=label))
_adaptive_selection_metrics = pd.concat(_adaptive_selection_metrics_parts, ignore_index=True)

validation_adaptive_selection, best_model_per_destination = build_adaptive_selection(MODEL_REGISTRY, _adaptive_selection_metrics, VALIDATION_MONTHS, REQUIRED_DESTINATIONS, training_actual_wide_to_2023M02, TRAIN_END)
MODEL_REGISTRY.append((validation_adaptive_selection, "adaptive_selection", "adaptive_selection"))

print("Model chosen per destination:")
best_model_per_destination
````

**Q6 Part A - MASE and MAPE**
Two views, not an option: MASE scales the error using the average monthly change from each destination, so it's comparable across series. MAPE expresses the error as a percentage of the actual demand, so it is easier to communicate. MAE is only used at the destination level as the MASE numerator only. And never averaged across the destinations because the 20 series differ by orders of magnitude.

MASE is scale-free. In the validation window, Hawaii never exceeds a few thousand arrivals a month while Hong Kong China stays above 1.97 million. Raw MAE cannot rank these destinations, but MASE can. This is why mean_mase without the weight of baseline_validation_summary and model_summary is meaningful. MASE under 1 also means the forecast beats the historical average one-step change.

MASE denominator can be unstable. Our denominators use all valid lag-1 pairs up to 2023M02; they mix pre-2020 growth, the COVID collapse, and the 2022-2023 reopening. All 20 destinations return mase_available=True and denominator_warning=False, with 115 (Chile) to 409 (Australia) valid pairs. But for destinations such as Taiwan China, and denominator is far larger than recent month-to-month movements. So the naive lag01 baseline scores MASE 0.280 while its MAPE on the same row is 37.1%.

MAPE is not suitable for zero or small actual values. Zero actual values will be deleted to avoid dividing by zero. In 2023M03-2023M07, there is no actual, so the row of data will be missing. small actual will be distortion. 

So, naive_lag12 is the weakest baseline (MASE 4,410; MAPE 77.9%) because it only repeats 2022 months when outbound travel was still largely closed. naive_lag1 is better (2.671/48.1%) but flat across all five validation months.
